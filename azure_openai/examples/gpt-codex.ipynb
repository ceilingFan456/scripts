```python
import requests
import json
import os

def call_azure_responses_api(messages, model="gpt-5.3-codex"):
    # 1. Configuration - adjust if your .sh file uses a different variable name
    url = "https://e0271-miptdstj-eastus2.cognitiveservices.azure.com/openai/responses?api-version=2025-04-01-preview"
    api_key = ""
    
    # 2. Flatten the messages list into a single "input" string
    # This mimics the Chat ML format the model expects
    flattened_input = ""
    for msg in messages:
        role = msg['role'].capitalize()
        content = msg['content']
        flattened_input += f"{role}: {content}\n"
    
    # 3. Prepare the payload with the updated parameter names
    headers = {
        "Content-Type": "application/json",
        "Authorization": f"Bearer {api_key}"
    }
    
    data = {
        "model": model, ## or gpt-5.2-codex
        "input": flattened_input.strip(),
        "max_output_tokens": 16384
    }

    # 4. Execute the request
    response = requests.post(url, headers=headers, json=data)
    
    if response.status_code == 200:
        return response.json()
    else:
        return f"Error {response.status_code}: {response.text}"

# --- Example Usage ---
chat_history = [
    {"role": "user", "content": "I am working on pathology AI research."},
    {"role": "assistant", "content": "That sounds fascinating! Are you focusing on Whole Slide Images (WSIs)?"},
    {"role": "user", "content": "Yes, specifically liver cancer slides. Can you suggest a Python library for tiling WSIs?"}
]

result = call_azure_responses_api(chat_history)
print(json.dumps(result, indent=2))
```
